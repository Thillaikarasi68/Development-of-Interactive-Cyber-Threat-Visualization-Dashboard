STEP 3: Clean the Dataset

```python
# Remove duplicate records
df = df.drop_duplicates()

# Standardize column names
df.columns = df.columns.str.lower()

df.head()
```

 STEP 4: BASIC EXPLORATORY DATA ANALYSIS (EDA)

```python
attacks_per_year = df.groupby('year').size().reset_index(name='attack_count')
attacks_per_year
```

Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,4))
sns.barplot(x='year', y='attack_count', data=attacks_per_year)
plt.title('Cyber Attacks Per Year')
plt.xlabel('Year')
plt.ylabel('Number of Attacks')
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.plot(attacks_per_year['year'],
         attacks_per_year['attack_count'],
         marker='o')

plt.title('Trend of Cyber Attacks (2016–2025)')
plt.xlabel('Year')
plt.ylabel('Number of Attacks')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.fill_between(attacks_per_year['year'],
                 attacks_per_year['attack_count'],
                 alpha=0.7)

plt.title('Growth of Cyber Attacks Over Years')
plt.xlabel('Year')
plt.ylabel('Number of Attacks')
plt.show()
```

STEP 5: Attack Type Analysis

```python
attack_types = df['type_of_attack'].value_counts().reset_index()
attack_types.columns = ['attack_type','count']
attack_types
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='count', y='attack_type', data=attack_types)
plt.title('Most Common Cyber Attacks')
plt.show()

```

STEP 6: Location-wise Analysis

```python
location_attacks = df['location'].value_counts().reset_index()
location_attacks.columns = ['location','count']
location_attacks
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='count', y='location', data=location_attacks)
plt.title('Cyber Attacks by Location')
plt.show()
```

STEP 7: Device Used for Attacks

```python
device_attacks = df['attack_device'].value_counts().reset_index()
device_attacks.columns = ['device','count']
device_attacks
```

```python
plt.figure(figsize=(9,5))
sns.barplot(x='count', y='device', data=device_attacks)
plt.title('Devices Used in Cyber Attacks')
plt.show()
```

STEP 8: Security Mechanism Analysis

```python
security_counts = df['security_type'].value_counts()
security_counts
```

```python
plt.figure(figsize=(8,8))
plt.pie(security_counts, labels=security_counts.index, autopct='%1.1f%%')
plt.title('Security Mechanisms Distribution')
plt.show()
```
